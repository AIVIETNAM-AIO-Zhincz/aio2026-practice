## 1. Tạo database trong bộ nhớ + bảng

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.executescript("""
CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    first_name  TEXT,
    last_name   TEXT,
    state       TEXT,
    points      INTEGER
);
CREATE TABLE orders (
    order_id    INTEGER PRIMARY KEY,
    customer_id INTEGER,
    order_date  TEXT
);
""")
print("Tao bang xong")
```

## 2. INSERT dữ liệu

```python
cur.executemany(
    "INSERT INTO customers VALUES (?,?,?,?,?)",
    [
        (1, "Babara", "MacCaffrey", "MA", 2273),
        (2, "Ines", "Brushfield", "VA", 947),
        (3, "Freddi", "Boagey", "CO", 2967),
    ],
)
cur.executemany(
    "INSERT INTO orders VALUES (?,?,?)",
    [
        (1, 1, "2024-01-01"),
        (2, 3, "2024-01-02"),
    ],
)
conn.commit()
```

## 3. SELECT + WHERE + ORDER BY + LIMIT

```python
for row in cur.execute("""
    SELECT * FROM customers
    WHERE points > 1000
    ORDER BY points DESC
    LIMIT 2
"""):
    print(row)
```

## 4. IN / BETWEEN

```python
q = """SELECT first_name, state, points FROM customers
       WHERE state IN ('VA','CO') AND points BETWEEN 500 AND 3000"""
for row in cur.execute(q):
    print(row)
```

## 5. JOIN hai bảng qua khoá chung

```python
q = """SELECT c.first_name, o.order_id, o.order_date
       FROM customers c JOIN orders o ON c.customer_id = o.customer_id"""
for row in cur.execute(q):
    print(row)
conn.close()
```
