## 1. `for` và `range()`; duyệt iterable

```python
for i in range(1, 7, 2):  # 1, 3, 5
    print(i, end=" ")
print()
for ch in "Hello":
    print(ch, end=" ")
```

## 2. Biến tích luỹ (accumulator)

```python
total = 0
for i in range(5):
    total += i
print("tong 0..4 =", total)  # 10

result = 1
for i in range(1, 5):
    result *= i
print("4! =", result)  # 24
```

## 3. `break` và `continue`

```python
for i in range(10):
    if i == 5:
        break
    print(i, end=" ")
print("| break tai 5")

for i in range(10):
    if i == 5:
        continue
    print(i, end=" ")
print("| continue bo qua 5")
```

## 4. `while` và `while True ... break`

```python
import random

i = 0
while i < 5:
    print(i, end=" ")
    i += 1
print()

while True:
    num = random.randint(0, 10)
    if num == 5:
        print("Ra 5, dung!")
        break
```

## 5. Ví dụ: ước lượng Pi, e; căn bậc hai Newton

```python
import math


def estimate_pi(n=1000):
    pi = 0
    for i in range(1, n):
        pi += (-1) ** (i + 1) / (2 * i - 1)
    return pi * 4


def estimate_e(n=20):
    e, fact = 1.0, 1
    for i in range(1, n + 1):
        fact *= i
        e += 1 / fact
    return e


def sqrt_newton(a, n=5):
    r = a / 2.0
    for _ in range(n):
        r = (r + a / r) / 2.0
    return r


print("Pi ~", estimate_pi(), "vs", math.pi)
print("e  ~", estimate_e(), "vs", math.e)
print("sqrt(9) =", sqrt_newton(9))
```

## 6. Diện tích hình tròn đơn vị bằng tổng Riemann

```python
import math

delta_x = 0.01
n = int(2 / delta_x)
x, half_area = -1.0, 0.0
for _ in range(n):
    half_area += delta_x * math.sqrt(max(0, 1 - x * x))
    x += delta_x
print("Dien tich ~", half_area * 2, "vs pi =", math.pi)
```
