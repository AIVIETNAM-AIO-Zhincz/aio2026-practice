## 1. Biến và 4 kiểu dữ liệu cơ bản

```python
name = 'AI VIETNAM'   # String
age = 21              # Integer
height = 1.72         # Float
is_student = True     # Boolean
for v in (name, age, height, is_student):
    print(v, '->', type(v))
```

## 2. Hàm và rẽ nhánh `if-elif-else`

```python
def classify(score):
    if score >= 8:
        return 'Giỏi'
    elif score >= 6.5:
        return 'Khá'
    elif score >= 5:
        return 'Trung bình'
    else:
        return 'Yếu'

for s in [9, 7, 5.5, 3]:
    print(s, '->', classify(s))
```

## 3. ReLU — ví dụ của `if`
$\text{ReLU}(x) = \max(0, x)$

```python
def relu(x):
    result = 0
    if x > 0:
        result = x
    return result

data = [1, 5, -4, 3, -2]
print([relu(x) for x in data])   # [1, 5, 0, 3, 0]
```

## 4. Rule-based Chatbot (cây if-else cố định)

```python
def chatbot(msg):
    msg = msg.lower()
    if 'xin chào' in msg or 'hello' in msg:
        return 'Xin chào quý khách! Tôi có thể giúp gì?'
    elif 'bảo hành' in msg:
        return 'Quý khách tra cứu bằng SĐT hay số seri ạ?'
    elif 'mua' in msg:
        return 'Quý khách muốn mua sản phẩm nào ạ?'
    else:
        return 'Để tôi kết nối nhân viên hỗ trợ.'

print(chatbot('Hello'))
print(chatbot('Tôi cần bảo hành'))
```

## 5. `match-case` (Python >= 3.10) — thay if-else dài

```python
def schedule(time):
    match time:
        case '5 AM':
            return 'Wake up'
        case '6 AM':
            return 'Yoga'
        case _:
            return 'Do something else'

print(schedule('6 AM'), '|', schedule('9 AM'))
```
