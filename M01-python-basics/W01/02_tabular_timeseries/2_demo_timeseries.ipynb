# Demo Time-series đơn giản với Pandas


1. Đọc dữ liệu bằng `pandas`
2. Xem nhanh cấu trúc dữ liệu
3. Chuyển cột `Date` sang kiểu thời gian
4. Vẽ line chart doanh thu theo ngày


## 1. Import thư viện

```python
import matplotlib.pyplot as plt
import pandas as pd

%matplotlib inline
```

## 2. Tạo dataset demo và lưu thành file CSV

```python
data = {
    "Date": [
        "2026-05-01",
        "2026-05-02",
        "2026-05-03",
        "2026-05-04",
        "2026-05-05",
        "2026-05-06",
        "2026-05-07",
    ],
    "Revenue_M": [52, 61, 58, 65, 63, 67, 64],
    "Orders": [82, 95, 90, 101, 98, 106, 100],
    "Visitors": [1200, 1380, 1310, 1450, 1420, 1500, 1460],
}

demo_df = pd.DataFrame(data)
demo_df.to_csv("demo_timeseries_sales_short.csv", index=False)
print("Đã tạo file: demo_timeseries_sales_short.csv")
```

## 3. Đọc dữ liệu bằng Pandas

```python
df = pd.read_csv("demo_timeseries_sales_short.csv")
df.head()
```

## 4. Kiểm tra nhanh dữ liệu

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.info()
```

## 5. Kiểm tra missing value

```python
df.isna().sum()
```

## 6. Chuyển cột Date sang datetime

```python
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date")
df.head()
```

## 7. Vẽ line chart doanh thu theo ngày

```python
plt.figure(figsize=(10, 5))
plt.plot(df["Date"], df["Revenue_M"], marker="o", linewidth=2)
plt.title("Daily Revenue Time-series")
plt.xlabel("Date")
plt.ylabel("Revenue (Million VND)")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
